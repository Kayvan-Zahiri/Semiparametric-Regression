```python
import pandas as pd

df = pd.read_csv("./data/boston_housing.csv")
df.head()
```

```python
df.describe()
```

```python
boston_mapping = {
    "rm": "Average Number of Rooms",
    "crim": "Crime Rate Per Capita",
    "nox": "Nitric Oxides Concentration (pp10m)",
    "age": "Proportion of Units Built Prior to 1940",
    "dis": "Distance from Employment Centers",
    "rad": "Accessibility to Highways",
    "tax": "Property Tax Rate",
    "ptratio": "Pupil-Teacher Ratio",
}

[{"label": boston_mapping[key], "value": key} for key in boston_mapping.keys()]
```

```python
import plotly.express as px
value = "rm"
label = "Average number of rooms per dwelling"
fig = px.scatter(df, x=value, y="medv", trendline="ols")
fig.update_traces(
    marker=dict(size=6, color="gray", opacity=0.6), selector=dict(mode="markers")
)
fig.update_traces(line=dict(color="red"), selector=dict(mode="lines"))
fig.update_layout(
    title={
        "text": f"{label} Impact on Median House Price",
        "x": 0.5,
        "xanchor": "center",
    },
    hovermode="closest",
)
results = px.get_trendline_results(fig)
model_results = results.px_fit_results.iloc[0]
slope = model_results.params[1]
intercept = model_results.params[0]
equation = f"Equation of the fitted model is: y = {intercept:.2f} + {slope:.2f}x"
fig.show()
print(equation)
```

```python
import statsmodels.formula.api as smf

model = smf.ols(
    "medv ~ crim + dis + nox + rm + age + rad + tax + ptratio",
    df,
).fit()
model.summary()
```

```python
model = smf.ols("medv ~ crim + dis + nox + rm + age + rad + ptratio + lstat", df).fit()
model.summary()
```

```python
model = smf.ols("medv ~ age", df).fit()
model.summary()
```
